```python
import numpy as np
from glob import glob
from skimage import io
from os import path
import pandas as pd
```

```python
# Deconvolution Lab 2 Functions
DL2_PATH = path.join(path.expanduser('~'), 
    'Code','DeconvolutionLab2', 'target', 'DeconcolutionLab2_-0.1.0-SNAPSHOT.jar')
RL_Niter = 10 # Number of RL iterations

# jarname = 'DeconcolutionLab2_-0.1.0-SNAPSHOT.jar'

print(DL2_PATH)
```

```python
# Handle the batched image files to be processed:
PSF_file = {}
PSF_file['G'] = '/Users/xies/Library/CloudStorage/OneDrive-Stanford/Skin/PSFs/PSF_BW_525nm_sq_1024_112.tif' # Green
PSF_file['B'] = '/Users/xies/Library/CloudStorage/OneDrive-Stanford/Skin/PSFs/PSF_BW_460nm_sq_1024_112.tif' # Blue
PSF_file['R'] = '/Users/xies/Library/CloudStorage/OneDrive-Stanford/Skin/PSFs/PSF_BW_595nm_sq_1024_50.tif' # Red

channel = 'R'

dirname = '/Users/xies/Library/CloudStorage/OneDrive-Stanford/Skin/Two photon/NMS/Dermal topography/mTmG clones/05-29-2026 mTmG clone rep2/DOB 03-22-2026/F1 tail clip/R1'
flist = glob(path.join(dirname,f'**/{channel}_reg.tif'),recursive=True)
manifest = pd.DataFrame()
manifest['Filename'] = flist
manifest['Dirname'] = manifest['Filename'].apply(lambda x: path.dirname(x))
manifest['Outname'] = manifest['Dirname'].apply(lambda x: path.join(x,f'{channel}_decon.tif'))
manifest['Skip'] = False
# print(manifest['Filename'].values)

for idx,row in manifest.iterrows():

    input_file = row['Filename']
    output_file = row['Outname']
    if path.exists(output_file):
        print(f'{output_file} already exists...skipping.')
        manifest.loc[idx,'Skip'] = True
        continue

print(f'Total of {len(manifest)} files found. Skipping {manifest.Skip.sum()} existing ones.')
```
